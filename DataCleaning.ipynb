### Transforming data using a Function or Mapping

```python
data = pd.DataFrame({
    'Alpha' : ['A' , 'B' , 'C' , 'A' , 'D' , 'D' , 'A' , 'B' , 'C'] ,
    'Ounces' : np.arange(9)
})
```

```python
data
```

```python
cap_to_small = {
    'A' : 'a' ,
    'B' : 'b' ,
    'C' : 'c' ,
    'D' : 'd'
}
```

```python
data['Alpha_Small'] = data['Alpha'].map(cap_to_small)
```

```python
data
```

```python
def small_to_cap(x):
    return str(x).capitalize()

data['Alpha_Cap'] = data['Alpha_Small'].apply(small_to_cap)
```

```python
data
```
