### Computations with Categories

```python
draws = rng.standard_normal(1000)
```

```python
draws[:5]
```

```python
bins = pd.qcut(draws , 4)
```

```python
bins
```

```python
bins = pd.qcut(draws , 4 , labels = ['Q1' , 'Q2' , 'Q3' , 'Q4'])
```

```python
bins
```

```python
bins.codes[:10]
```

```python
bins = pd.Series(bins , name = 'quartile')
```

```python
bins.head()
```

```python
results = pd.Series(draws).groupby(bins , observed = False).agg(['count' , 'min' , 'max']).reset_index()
```

```python
results
```

```python
results['quartile']
```

```python
print("The 'quartile' column retains the original categorical info including the ordering")
```

#### Better performance with categoricals

```python
n = 10_000_000
```

```python
labels = pd.Series(['foo' , 'bar' , 'baz' , 'quz'] * int(n/4))
```

```python
categories = labels.astype('category')
```

```python
labels.memory_usage(deep = True)
```

```python
categories.memory_usage(deep = True)
```

```python
print('Categorical Significantly one fifth of default string types')
```

```python
%timeit labels.value_counts()
```

```python
%timeit categories.value_counts()
```

### Categorical Methods

```python
s = pd.Series(['a' , 'b' , 'c' , 'd'] * 2)
```

```python
cat_s = s.astype('category')
```

```python
cat_s
```

```python
cat_s.cat.codes
```

```python
cat_s.cat.categories
```

```python
act_cat = ['a' , 'b' , 'c' , 'd' , 'e']
```

```python
cat_s2 = cat_s.cat.set_categories(act_cat)
```

```python
cat_s2
```

```python
cat_s2.value_counts()
```

```python
cat_s.value_counts()
```

```python
print('While the data look unchanged the new categories will be reflected in the operations that use them')
```

```python
cat_s3 = cat_s[cat_s.isin(['a' , 'b'])]
```

```python
cat_s3
```

```python
cat_s3.cat.categories
```

```python
cat_s3.cat.remove_unused_categories()
```

```python
cat_s3.cat.categories
```

```python
cat_s3 = cat_s3.cat.add_categories('z')
```

```python
cat_s3.cat.categories
```

```python
cat_s3.cat.as_ordered()
```
