##Exercise 1.1 KNN classifier on shuffled MNIST data<br>
Run mnist1NNdemo and verify that the misclassification rate (on the first 1000 test cases) of MNIST of a
1-NN classifier is 3.8%. (If you run it all on all 10,000 test cases, the error rate is 3.09%.) Modify the code
so that you first randomly permute the features (columns of the training and test design matrices), as in
shuffledDigitsDemo, and then apply the classifier. Verify that the error rate is not changed

```python
import pandas as pd
import numpy as np

from sklearn.datasets import fetch_openml
```

```python
mnist = fetch_openml('mnist_784' , version = 1 , as_frame = True)
```

```python
X , y = mnist.data , mnist.target
```

```python
X.shape
```

```python
X.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors = 1)
```

```python
X_train = X[:60000]
X_test = X[60000:]

y_train = y[:60000]
y_test = y[60000:]
```

```python
knn.fit(X_train , y_train)
```

```python
y_preds = knn.predict(X_test)
```

```python
from sklearn import metrics

acc = metrics.accuracy_score(y_preds , y_test)
```

```python
acc
```

```python
misclassification_rate = (1-acc)*100
```

```python
np.round(misclassification_rate , 2)
```

```python
print('Misclassification Rate of 3.09 is verified')
```

```python
print("Now we will shuffle the dataset")
```

```python
X , y = mnist.data , mnist.target
```

```python
X.head()
```

```python
X = X.sample(frac = 1 , axis = 1)
```

```python
X.head()
```

```python
X_train = X[:60000]
X_test = X[60000:]

y_train = y[:60000]
y_test = y[60000:]
```

```python
model = KNeighborsClassifier(n_neighbors = 1)
```

```python
model.fit(X_train , y_train)
```

```python
y_preds = model.predict(X_test)
```

```python
acc = metrics.accuracy_score(y_preds , y_test)
```

```python
acc
```

```python
new_missclassification_rate = (1 - acc)*100
```

```python
np.round(new_missclassification_rate , 2)
```

```python
print('Hence Proved the missclassification rate remains same after shuffling the features or say columns')
```
