## Extension Data Types

Missing data handling for some numerical data types such as integers and Booleans was incomplete.<br> As a result, when missing data was introduced into such data, pandas converted the data type to float64 and used np.nan to represent null values.<br>
More recently, pandas has developed an extension type system allowing for new data types to be added even if they are not supported natively by NumPy

```python
import numpy as np
import pandas as pd
```

```python
s = pd.Series([1 , 2 , 3 , None])
```

```python
s
```

```python
s[3]
```

```python
s[3] is None
```

```python
s[3] is np.nan
```

```python
s.dtype
```

```python
print('Legacy behavior of using float64 data type and np.nan for missing values')
```

```python
s = pd.Series([1 , 2 , 3 , np.nan] , dtype = pd.Int64Dtype())
```

```python
s
```

```python
print('NA for missing value in extension type array')
```

```python
s[3]
```

```python
s[3] is pd.NA
```

```python
s[3] is np.nan
```

```python
s[3] is None
```

### String Functions in pandas

```python
data = {
    "Dave": "dave@google.com",
    "Steve": "steve@gmail.com",
    "Rob": "rob@gmail.com",
    "Wes": np.nan
    }
```

```python
data = pd.Series(data)
```

```python
data.isna()
```

```python
print('All string operation can be applied using .apply or .map but it stucks on nap values')
```

```python
at_positions = [x.find('@') for x in data]
```

```python
print("Due to presence of nan it raises error 'float' object has no attribute 'find'")
```

```python
at_positions = data.str.find('@')
```

```python
print('Hence str attribute is used')
```

```python
data.str.contains('@')
```

```python
print('As a result of the opertion the returned datatype of str.contains is object and not boolean')
```

```python
data_as_string = data.astype('string')
```

```python
data_as_string
```

```python
data_as_string.str.contains('@')
```

```python
print('Here we got boolean')
```

```python
pattern = r'([A-Z0-9._%+-]+)@([A-Z0-9.-]+)\.([A-Z]{2,})'
```

```python
data.str.findall(pattern , flags = re.IGNORECASE)
```

```python
matches = data.str.findall(pattern , flags = re.IGNORECASE).str[0]
```

```python
matches
```

```python
print('Ending str[0] gives the first match for each case')
```

```python
matches.str.get(1)
```

```python
data.str.extract(pattern , flags = re.IGNORECASE)
```
