```python
from sklearn.model_selection import train_test_split
```

```python
xtrain , xtest , ytrain , ytest = train_test_split(X , y , test_size = 0.2)
```

```python
class Naives_Bayes :
  def fit(self , X_train , y_train) :
    self.classes = np.unique(y_train)
    self.priors = [len(y_train[y_train == c]) / len(y_train) for c in self.classes]

    self.means = [X_train[y_train == c].mean() for c in self.classes]
    self.stds = [X_train[y_train == c].std() for c in self.classes]

  def likelihood_calc(self , row , class_idx) :
    elipson = 10e-9
    likelihood = 1
    for f in row.index :
      mean = self.means[class_idx][f]
      std = self.stds[class_idx][f] + elipson
      likelihood *= (1 / np.sqrt(2 * np.pi * std ** 2) * np.exp((-1) * ((row[f] - mean) ** 2) / (2 * std ** 2)))
    return likelihood

  def predict(self , X) :
    y_pred = []
    for _ , row in X.iterrows() :
      posterior = []
      for i in range(len(self.classes)) :
        likelihood = self.likelihood_calc(row , i)
        posterior.append(likelihood * self.priors[i])
      y_pred.append(self.classes[np.argmax(posterior)])
    return y_pred
```

```python
nb = Naives_Bayes()
```

```python
nb.fit(xtrain , ytrain)
```

```python
pred = nb.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
acc = accuracy_score(ytest , pred)
```
