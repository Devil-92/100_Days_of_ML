```python
df_train = df.sample(frac = 0.7 , random_state = 42)
df_valid = df.drop(df_train.index)

df_train.head()
```

```python
df_train.index
```

```python
max_ = df_train.max(axis = 0)
min_ = df_train.min(axis = 0)

df_train = (df_train - min_) / (max_ - min_)
df_valid = (df_valid - min_) / (max_ - min_)

X_train = df_train.drop('quality' , axis = 1)
X_valid = df_valid.drop('quality' , axis = 1)
y_train = df_train['quality']
y_valid = df_valid['quality']
```

```python
X_train.shape
```

```python
model = keras.Sequential([
    keras.Input(shape = (11 ,)) ,
    layers.Dense(units = 512 , activation = 'relu') ,
    layers.Dense(units = 512 , activation = 'relu') ,
    layers.Dense(units = 512 , activation = 'relu') ,
    layers.Dense(1) ,
])
```

```python
model.compile(
    optimizer = 'adam' ,
    loss = 'mae'
)
```

```python
history = model.fit(
    X_train , y_train ,
    validation_data = (X_valid , y_valid) ,
    batch_size = 256 ,
    epochs = 10
)
```

```python
history_df = pd.DataFrame(history.history)

history_df.head()
```



> You might think about the information in the training data as being of two kinds: signal and noise. The signal is the part that generalizes, the part that can help our model make predictions from new data. The noise is that part that is only true of the training data; the noise is all of the random fluctuation that comes from data in the real-world or all of the incidental, non-informative patterns that can't actually help the model make predictions. The noise is the part might look useful but really isn't.



```python
## The validation set gives us an estimate of the expected error on the unseen data
```

```python
## The training loss will go down -> Model may Learn Noise or Model may Learn or It may learn Signal
## The validation loss will go down -> Model will Always learn Signal
```

```python
## If both curves goes down the model learns signal and if it learns noise there is a gap
## The size of the gap tells how much noise is learnt
```

```python
## Ideally , we should learn all the signal and no noise --> Not Possible
## We need the model to learn more signal at the cost of learning more noise
## As long as the trade is in our favour -> validation loss will decrease
## At certain point the trade can turn against us -> validation loss will increase
```

```python
## Overfitting -> The training set is when the loss is not as low as it could be because the model learned too much noise
## Underfitting -> The training set is when the loss is not as low as it could be because the model hadn't learnt enough signal
```

### Early Stopping

```python
## We can stop the training whenever it seems the validation loss isn't decreasing anymore -> Early Stopping
```

> Training with early stopping also means we're in less danger of stopping the training too early, before the network has finished learning signal. So besides preventing overfitting from training too long, early stopping can also prevent underfitting from not training long enough. Just set your training epochs to some large number (more than you'll need), and early stopping will take care of the rest.
