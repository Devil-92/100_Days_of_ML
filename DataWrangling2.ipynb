## Reshaping and Pivoting

### Reshaping with Heirarchical Indexing

```python
data = pd.DataFrame(np.arange(6).reshape((2, 3)) ,
                    index = pd.Index(['Ohio', 'Colorado'], name = 'state') ,
                    columns = pd.Index(['one', 'two', 'three'], name = 'number'))
```

```python
data
```

```python
result = data.stack()
```

```python
result
```

```python
result.unstack()
```

```python
print("We can also unstack using levels")
```

```python
result.unstack(level = 0)
```

```python
result.unstack(level = 1)
```

```python
result.unstack(level = 'state')
```

```python
print('Unstacking might introduce missing data if all values are not present')
```

```python
s1 = pd.Series([0 , 1 , 2 , 3] , index = ['a' , 'b' , 'c' , 'd'] , dtype = 'Int64')
```

```python
s2 = pd.Series([4 , 5 , 6] , index = ['c' , 'd' , 'e'] , dtype = 'Int64')
```

```python
d = pd.concat([s1 , s2] , keys = ['one' , 'two'])
```

```python
d
```

```python
d.unstack()
```

```python
d.unstack().stack()
```

```python
d.unstack().stack(future_stack = True)
```

### Pivoting Long to Wide Format

```python
link = "https://raw.githubusercontent.com/wesm/pydata-book/refs/heads/3rd-edition/examples/macrodata.csv"
```

```python
data = pd.read_csv(link)
```

```python
data.head()
```

```python
data = data.loc[ : , ['year' , 'quarter' , 'realgdp' , 'infl' , 'unemp']]
```

```python
data.head()
```

```python
periods = pd.PeriodIndex(
    year=data.pop("year"),
    quarter=data.pop("quarter"),
    name = "date"
)
```

```python
periods
```

```python
data.index = periods.to_timestamp('D')
```

```python
data.head()
```

```python
data = data.reindex(columns = ['realgdp' , 'infl' , 'unemp'])
```

```python
data.columns.name = 'item'
```

```python
data.head()
```

```python
long_data = (data.stack().reset_index().rename(columns = {0 : 'value'}))
```

```python
long_data[:10]
```

```python
print('The is called long format for multiple time series and each row in the table represents a single observation')
```

Data is frequently stored this way in relational SQL databases, as a **fixed schema** (column names and data types) allows the number of distinct values in the item column to change as data is added to the table.

In the previous example, date and item would usually be the **primary keys** (in relational database parlance), offering both:
* Relational integrity
* Easier joins

In some cases, the data may be more difficult to work with in this format; you might prefer to have a DataFrame containing **one column per distinct item value** indexed by timestamps in the date column.

```python
pivoted = long_data.pivot(index = 'date' , columns = 'item' , values = 'value')
```

```python
pivoted.head()
```

```python
long_data['value_2'] = np.random.normal(0 , 1 , size = len(long_data))
```

```python
long_data[:10]
```

```python
pivoted = long_data.pivot(index = 'date' , columns = 'item')
```

```python
pivoted.head()
```

```python
print('Calling pivot is eqivalent to creating a heirarchical index using set_index followed by a call to unstack')
```

```python
long_data.head()
```

```python
unstack = long_data.set_index(['date' , 'item']).unstack(level = 1)
```

```python
unstack.head()
```

### Pivoting Wide to Long Format

```python
print('It is inverse of long to wide and done by melt inplace of pivot')
```

```python
print('Rather than transforming one column into many in a new df it merges many columns into one')
```

```python
df = pd.DataFrame({
    'key': ['foo' , 'bar' , 'baz'] ,
    'A': [1 , 2 , 3] ,
    'B': [4 , 5 , 6] ,
    'C': [7 , 8 , 9]
})
```

```python
df
```

```python
print('We have to specify the group indicator and data values')
```

```python
melted = pd.melt(df , id_vars = 'key')
```

```python
melted
```

```python
reshaped = melted.pivot(index = 'key' , columns = 'variable' , values = 'value')
```

```python
reshaped
```

```python
print('Since the pivot creates an index from the column used as row labels , we can use reset_index to move the data back into column')
```

```python
reshaped.reset_index()
```

```python
print('We can specify the subset of columns to use as val')
```

```python
pd.melt(df , id_vars = 'key' , value_vars = ['A' , 'B'])
```

```python
print("pandas.melt can be used without group identifiers")
```

```python
pd.melt(df)
```

```python
pd.melt(df , value_vars = ['A' , 'B' , 'C'])
```
