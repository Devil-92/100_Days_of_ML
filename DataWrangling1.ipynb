### Reordering and Sorting Levels

```python
print('swaplevel method takes two level numbers or names and returns a new object with levels interchanged')
```

```python
frame.swaplevel('key1' , 'key2')
```

```python
print('sort_index by default sorts the data using all index levels but passing a single level or subset of levels sort that many level')
```

```python
frame.sort_index(ascending = False)
```

```python
frame.sort_index(level = 1)
```

```python
frame.swaplevel(0 , 1).sort_index(level = 0)
```

```python
print('for sorting use sort_index(level = x) and for reordering we use swaplevel')
```

```python
print('Data selection is better if the index is sorted lexographically and we it by calling sort_index')
```

```python
df1
```

```python
df2
```

```python
print('pd merges on overlapping column name as key but it is good to specify')
```

```python
df3 = pd.DataFrame({
    'key1' : ['b' , 'b' , 'a' , 'c' , 'a' , 'a' , 'a'] ,
    'data1' : pd.Series(range(7))
})
```

### Combining Data with Overlap

```python
a = pd.Series([np.nan , 2.5 , 0 , 3.5 , 4.5 , np.nan] , index = ['f' , 'e' , 'd' , 'c' , 'b' , 'a'])
```

```python
a
```

```python
b = pd.Series([0 , np.nan , 2 , np.nan , np.nan , 5] , index = ['a' , 'b' , 'c' , 'd' , 'e' , 'f'])
```

```python
b
```

```python
np.where(pd.isna(a) , b , a)
```

```python
print('Here our primary objective is to patch the null values in a by non null b values and the index are not aligned by np.where')
```

```python
print('Hence for patching combine_first is better')
```

```python
a.combine_first(b)
```

```python
print('Same patching the missing data happens in DataFrames column by column')
```

```python
df1 = pd.DataFrame({
    'a' : [1 , np.nan , 5 , np.nan] ,
    'b' : [np.nan , 2 , np.nan , 6] ,
    'c' : range(2 , 18 , 4)
})
```

```python
df2 = pd.DataFrame({
    'a' : [5 , 4 , np.nan , 3 , 7] ,
    'b' : [np.nan , 3 , 4 , 6 , 8]
})
```

```python
df1
```

```python
df2
```

```python
df1.combine_first(df2)
```
